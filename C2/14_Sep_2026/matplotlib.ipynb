```python
import matplotlib.pyplot as plt
x=[10,15,20,25,30,35,40,45,50]
y=[1000,2000,3000,4000,5000,6000,7000,8000,9000]
plt.figure(figsize=(8,5))
plt.plot(x,y, color='red',marker='o',linestyle='--',linewidth=2,markersize=8)
plt.title("comparison of age VS salary")
plt.xlabel("Age")
plt.ylabel("salary")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[10,15,20,25,30,35,40,45,50]
y1=[1000,2000,3000,4000,5000,6000,7000,8000,9000]
y2=[3000,5000,7000,4000,6000,8000,10000,5000,7000]
y3=[2000,3000,8000,6000,4000,5000,8000,7000,10000]
# plt.figure(figsize=(8,5))
plt.plot(x,y1,label='1st 4 month profit')
plt.plot(x,y2,label='2nd 4 month profit')
plt.plot(x,y3,label='3rd 4 month profit')
plt.title("comparsion of sales profit")
plt.xlabel("Months")
plt.ylabel("Quater Sales profit")
plt.legend()
plt.show()
```
